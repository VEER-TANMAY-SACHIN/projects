Here's an example of how to use `extend()`:

```python
# Let's reset our list 'l' for a clear demonstration
l = [1, 2, 3, 4, 5]

# Create another list to extend 'l' with
another_list = [6, 7, 8]

print(f"Original list 'l': {l}")
print(f"List to extend with: {another_list}")

l.extend(another_list)
print(f"List 'l' after extend: {l}")

# You can also extend with other iterables, like a string (it adds each character)
l.extend('abc')
print(f"List 'l' after extending with a string: {l}")
```
